# Scenario comparison for infra-enshure coupling meeting

This notebook compares Austria (AT) results across PyPSA-Eur scenarios for
marginal prices and optimal installed capacities.

```python
import pypsa
import eval_helpers as eh

input_dict = {
    "scenario_name_1": "path_to_scenario_file_1.nc",
    "scnenario_name_2": "path_to_scenario_file_2.nc",
    "AT_more_ambition_net_zero_2040": "~/Documents/infra-enshure/202606-coupling_workshop/mini_model_emissions_budget/AT_more_ambition_net_zero_2040/networks/base_s_adm__none_2040.nc",
}

networks = eh.load_networks(input_dict)
list(networks)
```

## Marginal prices (EUR/MWh)

```python
df_electricity = eh.compare_prices(
    networks=networks,
    bus_carriers=eh.PRICE_CATEGORIES['Electricity'],
    country_prefix='AT',
)
fig_electricity = eh.plot_grouped_bars(
    df_electricity,
    title='Electricity Prices',
    y_axis_title='€/MWh',
)
display(df_electricity)
fig_electricity.show()
```

```python
df_heat = eh.compare_prices(
    networks=networks,
    bus_carriers=eh.PRICE_CATEGORIES['Heat'],
    country_prefix='AT',
)
fig_heat = eh.plot_grouped_bars(
    df_heat,
    title='Heat Prices',
    y_axis_title='€/MWh',
)
display(df_heat)
fig_heat.show()
```

```python
df_h2 = eh.compare_prices(
    networks=networks,
    bus_carriers=eh.PRICE_CATEGORIES['H2'],
    country_prefix='AT',
)
fig_h2 = eh.plot_grouped_bars(
    df_h2,
    title='H2 Prices',
    y_axis_title='€/MWh',
)
display(df_h2)
fig_h2.show()
```

## Installed capacity (GW)

```python
df_electricity_capacity_generators = eh.compare_installed_capacity(
    networks=networks,
    components=['Generator'],
    carriers=eh.ELECTRICITY_GENERATOR_CARRIERS,
    bus_carriers=eh.ELECTRICITY_BUS_CARRIERS,
    country_prefix='AT',
)
fig_electricity_capacity_generators = eh.plot_grouped_bars(
    df_electricity_capacity_generators,
    title='Electricity Installed Capacity — Generators',
    y_axis_title='GW',
)
display(df_electricity_capacity_generators)
fig_electricity_capacity_generators.show()
```

```python
df_heat_capacity_generators = eh.compare_installed_capacity(
    networks=networks,
    components=['Generator'],
    carriers=eh.HEAT_GENERATOR_CARRIERS,
    bus_carriers=eh.HEAT_BUS_CARRIERS,
    country_prefix='AT',
)
fig_heat_capacity_generators = eh.plot_grouped_bars(
    df_heat_capacity_generators,
    title='Heat Installed Capacity — Generators',
    y_axis_title='GW',
)
display(df_heat_capacity_generators)
fig_heat_capacity_generators.show()
```
